## Feature Engineering - Standarization


```python
import pandas as pd
import numpy as np 
import seaborn as sns 
import matplotlib.pyplot as plt
```

```python
# importing The DataSet 

df=pd.read_csv('../datasets/Social_Network_Ads.csv')
df.sample(1) # Okay Its Working Correctly 
```

```python
df=df[["Age","EstimatedSalary","Purchased"]]
df.sample(2)  ## getting the Only useful Columns 
# other way to do that is 
# df=df.iloc[:,2:]
# df.sample(1)

X=df[["Age","EstimatedSalary"]]
# print(X.sample(2)) ## just for Checking 
Y=df["Purchased"]
# print(Y.sample(2)) 

```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.3)

X_train.shape # (280 , 2)
X_test.shape  # (180 , 2)
```

```python
from sklearn.preprocessing import StandardScaler

scaler=StandardScaler()
scaler.fit(X_train)

# agr fit na lagae to best way 
# X_train_scaled=scaler.fit_transform(X_train) ya wali lien aik sath 2 kam kar leti ha 

X_train_scaled=scaler.transform(X_train)
# print(X_train_scaled)

X_test_scaled=scaler.transform(X_test)
# print(X_test_scaled)

scaler.mean_
```

```python
X_train_scaled=pd.DataFrame(X_train_scaled,columns=X_train.columns)
print(X_train_scaled.sample(5))
```

```python
from sklearn.linear_model import  LogisticRegression

lr=LogisticRegression()

lr.fit(X_train_scaled,Y_train)


y_prediction = lr.predict(X_test_scaled)
```

```python
from sklearn.metrics import accuracy_score

model_accuracy = accuracy_score(Y_test, y_prediction)

print(f'The Model accuracy is {model_accuracy * 100:.2f}%')
```
